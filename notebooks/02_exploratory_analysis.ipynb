# UrbanTwin • Milestone 2: Exploratory Data Analysis (EDA)
**Module**: CongestAir (Historical Air Quality in Pune, India)

This notebook performs in-depth statistical exploration and seasonal profiling:
- Descriptive statistics (parametric and non-parametric) for all pollutants
- Inter-pollutant correlation analysis
- Long-term trends for AQI, PM2.5, PM10, and NO2
- Dominant pollutant frequencies
- Seasonality (monthly and inter-annual variations)
- Spatial distribution comparison across monitoring stations

```python
import sys
from pathlib import Path
project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import pandas as pd
import matplotlib.pyplot as plt
from pipelines.data_quality import DataQualityAnalyzer, POLLUTANT_COLUMNS

csv_path = project_root / 'data' / 'all_stations_combined.csv'
df = pd.read_csv(csv_path)
analyzer = DataQualityAnalyzer(df)
clean_df = analyzer.df
```

## 1. Pollutant Descriptive Statistics
Parametric measures (mean, std), non-parametric quantiles (min, 25%, median, 75%, max), and distributional moments (skewness, kurtosis).

```python
stats_df = analyzer.compute_statistics()
stats_df
```

## 2. Correlation Matrix Between Pollutants
Pairwise Pearson correlation coefficients between pollutants and AQI.

```python
corr = clean_df[POLLUTANT_COLUMNS].corr()
corr
```

## 3. Dominant Pollutant Frequencies
Distribution of primary drivers of the Indian National Air Quality Index (NAAQI).

```python
clean_df['dominant_pollutant'].value_counts(dropna=False)
```

## 4. Seasonal and Monthly AQI Profiling
Evaluates winter spikes versus monsoon washout patterns.

```python
monthly_profile = clean_df.groupby('month')['aqi'].agg(['count', 'mean', 'median', 'std']).round(2)
monthly_profile
```

## 5. Station-Wise AQI Comparison
Compares pollution severity across different monitoring localities in Pune.

```python
clean_df.groupby('station_id')['aqi'].describe().round(2)
```
